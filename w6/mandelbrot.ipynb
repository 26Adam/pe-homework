```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib ipympl
from IPython.display import SVG, display

def mandelbrot(h, w):
    x = np.linspace(-2, 0.8, w)
    y = np.linspace(-1.4, 1.4, h)
    X, Y = np.meshgrid(x, y)
    diverge = np.zeros((h, w))
    for i in range(h):
        for j in range(w):
            c = X[i, j] + 1j * Y[i, j]
            z = c
            for n in range(20):
                z = z**2 + c
                if abs(z) > 2:
                    diverge[i, j] = n + 1
                    break
    plt.imshow(diverge)
    plt.show()
    return diverge    
diverge = mandelbrot(100, 100)

def mandelbrot2(h, w, max_iter, fichier):
    x = np.linspace(-2, 0.8, w)
    y = np.linspace(-1.4, 1.4, h)
    X, Y = np.meshgrid(x, y)
    diverge = np.zeros((h, w))
    for i in range(h):
        for j in range(w):
            c = X[i, j] + 1j * Y[i, j]
            z = c
            for n in range(max_iter):
                z = z**2 + c
                if abs(z) > 2:
                    diverge[i, j] = n + 1
                    break
    plt.imshow(diverge)
    plt.savefig(fichier)
    return diverge
    
#on pourrait enregistrer en JPEG (image) ou SVG (image vectorielle)

mandelbrot2(100, 100, 20, "test1.svg")
display(SVG(filename="test1.svg"))
mandelbrot2(500, 500, 20, "test2.svg")
display(SVG(filename="test2.svg"))
mandelbrot2(100, 100, 50, "test3.svg")
display(SVG(filename="test3.svg"))
```
